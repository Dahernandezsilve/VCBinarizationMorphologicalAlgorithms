## Comparación entre algoritmos 📊

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
niblack = pd.read_csv('resultadosNiblack.csv')
otsu = pd.read_csv('resultadosOtsu.csv')

display(niblack.head())
display(otsu.head())
```

```python
df_metrics = pd.DataFrame({
    'Accuracy': niblack['Accuracy'],
    'Recall': niblack['Recall'],
    'Precision': niblack['Precision'],
    'F1 Score': niblack['F1 Score']
})

df_metrics2 = pd.DataFrame({
    'Accuracy': otsu['Accuracy'],
    'Recall': otsu['Recall'],
    'Precision': otsu['Precision'],
    'F1 Score': otsu['F1 Score']
})

plt.figure(figsize=(8, 6))
sns.heatmap(df_metrics.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlación entre Métricas del Algoritmo Niblack')
plt.show()

plt.figure(figsize=(8, 6))
sns.heatmap(df_metrics2.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlación entre Métricas del Algoritmo Otsu')
plt.show()
```

```python
metricas = ['Accuracy', 'Recall', 'Specificity', 'Precision', 'F1 Score']

categorias = metricas
niblack_values = niblack[metricas].mean()
otsu_values = otsu[metricas].mean()

# Número de métricas
N = len(categorias)

# Ángulos para las categorías
angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()

# Completar el círculo
niblack_values = np.concatenate((niblack_values, [niblack_values[0]]))
otsu_values = np.concatenate((otsu_values, [otsu_values[0]]))
angles += angles[:1]

# Graficar
plt.figure(figsize=(8, 8))
plt.subplot(111, polar=True)
plt.plot(angles, niblack_values, label='Algoritmo Niblack', linewidth=2, linestyle='solid')
plt.plot(angles, otsu_values, label='Algoritmo Otsu', linewidth=2, linestyle='solid')

plt.fill(angles, niblack_values, color='blue', alpha=0.25)
plt.fill(angles, otsu_values, color='orange', alpha=0.25)

plt.title('Comparación de Métricas entre Algoritmos')
plt.xticks(angles[:-1], categorias, fontsize=12)
plt.legend(loc='upper right')
plt.show()
```
